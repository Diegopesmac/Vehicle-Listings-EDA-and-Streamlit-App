# EDA - Vehicle Listings
Este notebook demonstra visualizações básicas com Plotly Express usando o arquivo `vehicles.csv`.

- Histograma da coluna `odometer`
- Gráfico de dispersão `price` vs `odometer`

```python
import pandas as pd
import plotly.express as px

# Carregar dados
df = pd.read_csv('../vehicles.csv')
df.head()

# Histograma - odômetro
if 'odometer' in df.columns:
    fig_hist = px.histogram(df, x='odometer', nbins=30, title='Distribuição do odômetro')
    fig_hist.show()
else:
    print('Coluna odometer não encontrada')

# Dispersão - preço vs odômetro
color_col = 'model' if 'model' in df.columns else ('manufacturer' if 'manufacturer' in df.columns else None)
if {'price','odometer'}.issubset(df.columns):
    fig_scatter = px.scatter(df, x='odometer', y='price', color=color_col, title='Preço vs Odômetro')
    fig_scatter.show()
else:
    print('Colunas price e/ou odometer não encontradas')
```
